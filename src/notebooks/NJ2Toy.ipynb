```python
import skbio
import numpy as np
```

```python
from skbio import DistanceMatrix
```

```python
from skbio.tree import nj
```

```python
def generate_distance_matrix(size):
    # Create a random distance matrix
    random_matrix = np.random.rand(size, size)
    
    # Make the matrix symmetric
    distance_matrix = (random_matrix + random_matrix.T) / 2
    
    # Set the diagonal to zero
    np.fill_diagonal(distance_matrix, 0)
    
    return distance_matrix
```

```python
def print_distance_matrix(matrix):
    size = matrix.shape[0]
    print("Distance Matrix:")
    for i in range(size):
        for j in range(size):
            print(f"{matrix[i][j]:.2f}", end="\t")
        print()
```

```python
matrix_size = 20
```

```python
distance_matrix = generate_distance_matrix(matrix_size)
```

```python
print_distance_matrix(distance_matrix)
```
